```python
# Importing the matplotlib library
import matplotlib.pyplot as plt

# Creating data for the plot
x = [1, 2, 3, 4, 5]
y = [1, 4, 9, 16, 25]

# Plotting the data
plt.plot(x, y, marker='o', linestyle='-', color='b', label="y = x^2")

# Adding title and labels
plt.title("Simple Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")

# Adding a legend
plt.legend()

# Display the plot
plt.show()
```
